# HANDLE MISSING VALUES

```python
df.isnull().sum()
```

```python
df=df.dropna()
print(df.isnull().sum())
```

# REMOVE DUPLICATE VALUES

```python
print("Duplicates:",df.duplicated().sum())
df.drop_duplicates()
print("Duplicates after removing:",df.duplicated().sum())
```

# CORRECT INCONSITENT FORMATTING AND TEXT VALUES

```python
df.columns = df.columns.str.strip()   #remove extra space from column name
for col in df.select_dtypes(include="object"):   #remove extra space from textvalues
  df[col] = df[col].str.strip()
for col in df.select_dtypes(include="object"):      ##text to titlecase
  df[col] = df[col].str.title()
df.head()
```

```python
print(df.dtypes)
df["Years of Experience"] = df["Years of Experience"].astype(int)
```

```python
print(df.dtypes)
# Convert age column if present

if 'Age' in df.columns:
    df['Age'] = pd.to_numeric(df['Age'], errors='coerce')

# Convert salary column if present

if 'Salary' in df.columns:
    df['Salary'] = pd.to_numeric(df['Salary'], errors='coerce')

# Check updated types
print(df.dtypes)
```

# STANDARDIZE COLUMN NAMES

```python
# Convert column names to lowercase

df.columns = df.columns.str.lower()

# Replace spaces with underscores

df.columns = df.columns.str.replace(' ', '_')

# Remove special characters

df.columns = df.columns.str.replace('[^a-zA-Z0-9_]', '', regex=True)

print("Updated Column Names:")
print(df.columns)
```

# VALIDATE AND VERIFY

```python
print("Dataset Shape:")
print(df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)

print("\nPreview of Cleaned Dataset:")
df.head()
```

# EXPORT

```python
df.to_csv("cleaned_dataset.csv", index=False)
print("dataset export success")
```
