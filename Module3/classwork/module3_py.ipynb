# BAR CHARTS
CONSISTS OF CATEGORIAL DATA MEANS DATA IN CATEGORIES

```python
import matplotlib.pyplot as plt
import numpy as np

SUBJECT = ["PHYSICS","MATHS","HINDI","ENGLISH","CHEMISTRY", "MEDIN"]
MARKS = [70,60,55,88,90,75]
plt.figure(figsize=(8,6))

plt.bar(SUBJECT,MARKS,color="pink",edgecolor="yellow")
plt.xlabel("SUBJECT")
plt.ylabel("MARKS")

#title add
plt.title("MARKS OF STUDENTS")

#add value top of the bar
for index, value in enumerate (MARKS):
    plt.text(index,value+1,str(value),ha='center')

# Calculate the median of the marks
median_marks = np.median(MARKS)

# Add a horizontal line for the median
plt.axhline(median_marks,
            color='purple',
            linestyle='--',
            linewidth=2.3,
            label=f'Median: {median_marks}')

# Add a legend to show the median label
plt.legend()

# Tight layout
plt.tight_layout(pad=2.0)

# Show chart
plt.show()
```

# PIECHART
proportional data = slice,
slice name = label,
slice data = value,
(data distribution)
only 2 variable requird for piechart i.e; label,score

```python
import matplotlib.pyplot as plt
```

```python
Quaters = ["Q1","Q2","Q3","Q4"]
Profit = [40000,50000,50000,120000]
range=["pink","yellow","maroon","tan"]

# Calculate total profit for autopct function
total_profit = sum(Profit)

# Define a function to display numerical values
def autopct_func(pct):
    absolute_value = int(pct/100.*total_profit)
    return "{:.0f}".format(absolute_value)

#PLOT PIE CHART
plt.pie(Profit,labels=Quaters,
        autopct=autopct_func, # Use the custom function here
        startangle=40,          #startanlge=rotating
        explode=(0.1,0.3,0.5,0.7),  ##explode pie chart disperse(seperate)
        shadow=True,colors=range)

#TITLE ADD
plt.title("PROFIT vs QUATERS")

plt.axis("equal")     #to maintain the shape of piechart(i.e;circle)

plt.legend()
plt.show()
```

# SCATTER PLOTS
used to find relations

```python
NOF=[1,3,4,6,8]
Income=[1000,80,480,60,99]

#bubble add
Size=[100,400,1500,2800,7700]


colors=["maroon","tan","cyan","yellow","pink"]
#scatter plot
plt.scatter(NOF, Income,s=Size,c=colors,alpha=0.7)

plt.colorbar(label='Intensity')

#xaxis and yaxis data define
plt.xlabel("NOF")
plt.ylabel("INCOME")

#add title
plt.title("NOF vs INCOME")
plt.show()
```

```python
X1=[2,4,6,8]
X2=[1,3,5,7]
Y1=[11,13,15,17]
Y2=[12,14,16,18]
colors=["pink","maroon","yellow","purple"]


plt.scatter(X1,Y1,label="GROUP A",c=colors)
plt.scatter(X2,Y2,label="GROUP B",c=colors)

plt.xlabel("X-AXIS")
plt.ylabel("Y-AXIS")
plt.title("SCATTER PLOT")

plt.legend()
plt.show()
```
